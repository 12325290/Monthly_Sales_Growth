```python
import pandas as pd
df = pd.read_excel("Sales_Data_Raw.xlsx")

print(df.head())
```

```python
df = df.drop_duplicates()

df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df = df.dropna(subset=['Order Date'])

df['Profit'] = df['Sales'] - df['Cost']
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month

df.to_csv("Sales_Cleaned.csv", index=False)
```

```python
from sqlalchemy import create_engine

# Connection details
# Format: mysql+pymysql://username:password@localhost:port/database_name
# Note: If your password has '@', use '%40' instead.
engine = create_engine("mysql+pymysql://root:123456@localhost:3306/sales_db")

# Step 5: Load Data Frame into SQL Table [00:26:24]
df.to_sql(name='sales_data', con=engine, if_exists='replace', index=False)

print("Data successfully loaded into SQL!")
```
